```python
import pandas as pd
import evaluate
from comet import download_model, load_from_checkpoint
import os
```

```python
bleu = evaluate.load("sacrebleu")
```

```python
model_name = "Unbabel/wmt22-comet-da"
model_path = download_model(model_name)
comet_model = load_from_checkpoint(model_path)
```

```python
def eval_bl_cmt(csv_path, ref_path):
    df = pd.read_csv(csv_path)
    src_list = df["EnglishText"].tolist()
    hyp_list = df["TranslatedFrench"].tolist()
    with open(ref_path, 'r', encoding='utf-8') as f:
        ref_list = [line.strip() for line in f.readlines()]
    formatted_refs = [[r] for r in ref_list]
    bleu_results = bleu.compute(predictions=hyp_list, references=formatted_refs)
    print(f"BLEU Score: {bleu_results['score']:.4f}")
    comet_data = [{"src": s, "mt": h, "ref": r} for s, h, r in zip(src_list, hyp_list, ref_list)]
    comet_results = comet_model.predict(comet_data, batch_size=8, gpus=1 )
    print(f"COMET Score: {comet_results.system_score:.4f}")
```

Evaluation using a pre trained french emotion recogniser

```python
from transformers import pipeline
import torch
```

```python
pipe = pipeline("text-classification", model="astrosbd/french_emotion_camembert")
```

```python
import pandas as pd
import os

def analyze_emotion_file(csv_path, emotion_pipeline):
    df = pd.read_csv(csv_path)
    has_vad = 'Valence' in df.columns and 'Arousal' in df.columns

    texts_to_classify = [
        t if (pd.notna(t) and t.strip()) else "neutre" # Use a neutral word for empty rows
        for t in df["TranslatedFrench"]
    ]
    
    try:
        # Run the classifier in batches
        results = emotion_pipeline(texts_to_classify, batch_size=8, truncation=True)
    except Exception as e:
        print(f"ERROR during classification: {e}\n")
        return
        
    # The output is a list of dicts like {'label': 'joie', 'score': 0.99}
    df['DetectedEmotion'] = [res['label'] for res in results]
    print("Classification complete.\n")

    # --- 3. Analysis Part 1: Emotion Distribution ---
    print("--- Emotion Distribution ---")
    print("How frequent was each detected emotion?\n")
    
    emotion_counts = df['DetectedEmotion'].value_counts(normalize=True).mul(100)
    print(emotion_counts.head(10).to_markdown(floatfmt=".2f"))
    print("\n")

    # --- 4. Analysis Part 2: VAD Transfer ---
    if has_vad:
        print("---  VAD Transfer Analysis ---")
        print("What were the avg SOURCE VAD scores for each DETECTED emotion?\n")
        
        emotion_analysis = df.groupby('DetectedEmotion').agg(
            Avg_Source_Valence=('Valence', 'mean'),
            Avg_Source_Arousal=('Arousal', 'mean'),
            Avg_Source_Dominance=('Dominance', 'mean'),
            Count=('EnglishText', 'count')
        ).sort_values(by='Count', ascending=False)
        
        print(emotion_analysis.to_markdown(floatfmt=".2f"))
    else:
        print("--- VAD Transfer Analysis ---")
        print("Skipped. 'Valence'/'Arousal' columns not found in this CSV.")
    
```

```python
# --- Define paths ---
baseline_csv = "towerbase_output/dev_en2fr.csv" 
arousal_csv = "towerbase_output/dev_translations_arousal_eq3.csv"
novel_csv = "towerbase_output/dev_translations_novel_prompt.csv"

# --- Run the analysis, 
if os.path.exists(baseline_csv):
    analyze_emotion_file(baseline_csv, pipe) 
else:
    print(f"Skipping baseline, file not found: {baseline_csv}\n")
    
analyze_emotion_file(arousal_csv, pipe) 
analyze_emotion_file(novel_csv, pipe)   
```

```python
analyze_emotion_file("/data3/ritika_project/ml/towerbase_output/test_en2fr.csv", pipe)
analyze_emotion_file("/data3/ritika_project/ml/towerbase_output/test_translations_arousal_eq3.csv", pipe) 
analyze_emotion_file("/data3/ritika_project/ml/towerbase_output/test_translations_novel_prompt.csv", pipe) 
```
